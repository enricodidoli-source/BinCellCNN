```python
import pandas as pd
import numpy as np
import os

np.random.seed(42)

num_rows = 10000

# prob_class_1 in percentuale, diviso per 100
prob_class_1 = np.array([0.1, 0, 0.5, 0, 1, 0, 5, 0, 3, 0]) / 100
patients = [1, 1, 2, 2, 4, 4, 6, 6, 8, 10]


num_files = len(patients)

train_art_path = 'C:/Users/admin/BinCellCNN/Data/art_train_samples/'
os.makedirs(train_art_path, exist_ok=True)

y_path = f'{train_art_path}samples_labels/'
os.makedirs(y_path, exist_ok=True)

new_art_path = 'C:/Users/admin/BinCellCNN/Data/art_new_samples/'
os.makedirs(new_art_path, exist_ok=True)

for i in range(1, num_files + 1):
    pat = patients[i-1]
    p_1 = prob_class_1[i-1]
    
    data = {
        'marker_1': np.random.uniform(0, 3, num_rows),
        'marker_2': np.random.uniform(0, 1, num_rows),
        'marker_3': np.random.uniform(-1, 1, num_rows),
        'marker_4': np.random.uniform(0, 10, num_rows),
    }
    
    if p_1 > 0:
        data['class'] = np.random.choice(
            [0, 1], 
            size=num_rows, 
            p=[1 - p_1, p_1]
        )
    else:
        data['class'] = np.zeros(num_rows, dtype=int)
        
    df = pd.DataFrame(data)
    print(f"\n--- File {i} (Paziente {pat}) ---")
    print(df['class'].value_counts()) # Mostra quanti 0 e 1 ci sono
    
    if i < num_files - 1:
        save_path = train_art_path

        tot_positive = df['class'].sum()
        print('tot_positive', tot_positive)
            
        y = 1 if tot_positive > 0 else 0
        sample_y = {'label': [y]}
        sample_y_df = pd.DataFrame(sample_y)
            
        filename_y = f'{y_path}sample_y_{pat}_{i}.csv'
        sample_y_df.to_csv(filename_y, index=False)
        
    else:
        save_path = new_art_path
    
    filename = f'{save_path}sample_{pat}_{i}.csv'
    df.to_csv(filename, index=False)

print("\nTutti i file CSV sono stati generati con successo!")
```
